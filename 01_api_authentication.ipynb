# API Authentication

Every notebook in this repository starts by creating a *SapioUser*. That object holds the webservice URL and the credentials sent with each request.

The webservice API accepts the credentials below.

## API Users

Create an API user in App Setup -> User Manager -> Add User, and check the "API User Account" box.

An API user authenticates with a username and password. Create a separate API user for each integration. The username is written to the audit log for actions that user takes, so a separate user makes it possible to see which integration made a change. The user's permissions come from the user group you assign to that user.

Each concurrent connection made with these credentials consumes one concurrent connection seat.

Pass these fields to *SapioUser*:

* **url:** The address of the webservice. This is your system's URL followed by `/webservice/api`. On Sapio SaaS this looks like `https://customer_name.exemplareln.com/webservice/api`.
* **username:** The API user's username.
* **password:** The API user's password.
* **guid:** The GUID of the Sapio application when connecting to a dedicated host. More than one application can run on the same server, and the GUID selects which one. Locate it in the Sapio Informatics Portal, as shown in the screenshot below. On a SaaS URL, the application is determined from the URL, so `guid` can be left unset.
* **account_name:** The account name shown for that application in the portal, when the host requires it.

To find the account and portal information, navigate to the Sapio Portal and inspect the relevant application. In this example, the app GUID is "3c232543-f407-4828-aae5-b33d4cd31fa7".

![App guid in the Sapio portal](resources/portal_app_guid.png)

Replace the URL, GUID, account name, username, and password with the values for your system.

```python
from sapiopylib.rest.User import SapioUser

user = SapioUser(url="https://linux-vm:8443/webservice/api",
                 guid="your-app-guid",
                 account_name="your-account-name",
                 username="api_user",
                 password="password",
                 verify_ssl_cert=True)
```

Set `verify_ssl_cert` to `False` only when the server uses a self-signed certificate (such as a local development server). Leave it set to `True` when connecting to a server with a valid SSL/TLS certificate.

## API Key

Generate an API key from App Setup -> Manage Rules & Webhooks -> Webservice API.

An API key grants admin access to the system. Actions taken with the key are audited under a generic API user. Generating a new key invalidates the previous key. The key is shown when it is created and cannot be viewed again afterward.

Pass the key as `api_token`. Replace the URL and key with the values for your system.

```python
from sapiopylib.rest.User import SapioUser

user = SapioUser(url="https://linux-vm:8443/webservice/api",
                 api_token="your-api-key",
                 verify_ssl_cert=True)
```

## Bearer Token

A bearer token can authenticate webservice API requests when the Sapio system has OAuth configured. The token authorizes calls to the webservice API. It is separate from signing into the web client.

Configure the app's OAuth settings with an authorization server. That server issues the bearer token, and Sapio validates the token against it.

Pass the token as `bearer_token` on its own. If `username` and `password` are also provided, `SapioUser` will prioritize those credentials instead.

```python
from sapiopylib.rest.User import SapioUser

user = SapioUser(url="https://linux-vm:8443/webservice/api",
                 bearer_token="your-bearer-token",
                 verify_ssl_cert=True)
```
