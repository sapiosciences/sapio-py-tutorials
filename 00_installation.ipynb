## Welcome!
Welcome to the Sapio Python client tutorial.

This tutorial covers the installation and usage of the main Sapio client functions in Python. You will learn different ways to query the Sapio platform, manipulate records, edit ELN experiments, and host your own webhook server that receives context from Sapio and runs your custom logic.

## System Requirements
These libraries require **Python 3.13 or higher** and `pip` installed in your Python environment.

If your system has an older version of Python installed, you can keep it and run Sapio in an Anaconda environment that uses a newer Python install. In general, if you are not using a cloud environment such as Google Colab, you may find it easier to maintain your installation with Anaconda or Miniconda. [Please refer to their tutorials for details.](https://www.anaconda.com/)

If you are using Anaconda, make sure that pip is installed within the Anaconda environment and is not referencing the global Python environment.

In Linux, you can confirm which environment your pip belongs to by running:

    which pip
Look at the path in the output. If you are using Anaconda, it should point to your conda root. If it does not, you may need to install pip into your conda environment.

    conda install pip

## Installation

Sapio publishes two Python packages: **sapiopylib**, the SDK that wraps the platform's webservice (REST) API, and **sapiopycommons**, a utilities library built on top of sapiopylib and maintained by our own in-house developers. sapiopycommons depends on sapiopylib, so **you only need to install sapiopycommons** and sapiopylib will be installed with it. Only install sapiopylib on its own if you want the bare SDK without the utilities.

 - [sapiopylib on PyPI](https://pypi.org/project/sapiopylib/)
 - [sapiopycommons on PyPI](https://pypi.org/project/sapiopycommons/)

### Choosing a version

Both packages receive a new release for every major platform version, even if the webservice API did not change. This ensures package versions match your platform version. Install the sapiopycommons release that matches your system's version. For example, if your system is on 26.8, install sapiopycommons 38.0.1.26.8. You can find your version's release in the [sapiopycommons release history](https://pypi.org/project/sapiopycommons/#history).

We maintain backwards compatibility between versions, so it isn't strictly necessary to update the Python libraries when your platform is upgraded if active development on webhooks or webservice API integrations has stopped.

The `prior_releases/*` branches of this repository belong to older platform versions. These branches are no longer updated.

### Running the install

If you are launching from a Jupyter Notebook environment (this includes Amazon SageMaker Notebooks and Google Colab), your command is as follows:

    !pip install sapiopycommons==38.0.1.26.8

Remove the "!" prefix if you are running the command directly from a terminal:

    pip install sapiopycommons==38.0.1.26.8
