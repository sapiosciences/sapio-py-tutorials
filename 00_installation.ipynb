## System Requirements
Use of this library requires you to have installed **Python 3.13 or higher** as your Python environment, and it also requires you to have installed "pip" for Python environment.

If your system has an older version of Python installed, you may keep the older version of Python and sandbox Sapio inside Anaconda environment of another Python shell install. In general, if you are not using a cloud environment such as Google CoLab, you may find it easier to maintain your installation with Anaconda/Miniconda. [Please refer to their tutorials for details.](https://www.anaconda.com/)

If you are launching from Anaconda, check to make sure you have pip installed within the anaconda sandbox and it is not referencing the global Python environment.

In Linux, you can confirm your pip installation environment by typing.

    which pip
Look at the path in the output. If you are using Anaconda, this should point to your conda root. If it does not, you may need to install pip onto your conda environment.

    conda install pip

## Installation

Sapio publishes two Python packages: **sapiopylib**, the SDK that wraps the platform's webservice (REST) API, and **sapiopycommons**, a utilities library built on top of sapiopylib and maintained by our own in-house developers. sapiopycommons depends on sapiopylib, so **you only need to install sapiopycommons** and sapiopylib will be installed with it. Only install sapiopylib on its own if you want the bare SDK without the utilities.

 - [sapiopylib on PyPI](https://pypi.org/project/sapiopylib/)
 - [sapiopycommons on PyPI](https://pypi.org/project/sapiopycommons/)

### Choosing a version

Both packages receive a new release for every major platform version, even if the webservice API did not change. This is to reduce confusion as to which package version should be installed. Install the sapiopycommons release that matches your system's version. For example, if your system is on 26.8, install sapiopycommons 38.0.1.26.8. You can find your version's release in the [sapiopycommons release history](https://pypi.org/project/sapiopycommons/#history).

We maintain backwards compatibility between versions, so it isn't strictly necessary to update the Python libraries when your platform is upgraded if active development on webhooks or webservice API integrations has stopped.

The `prior_releases/*` branches of this repository belong to older platform versions. These branches are no longer updated.

### Running the install

If you are launching from a Juypter Notebook environment. (This would include Amazon SageMaker Notebooks and Google CoLab), your command will be as follows:

    !pip install sapiopycommons==38.0.1.26.8

Remove the "!" prefix in the command if you are launching directly from terminal, like so:

    pip install sapiopycommons==38.0.1.26.8
